```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# NOTE : le dataset n'est pas inclus dans ce depot (voir README).
# Remplacer le chemin ci-dessous par une image locale avant execution.
image_path = "dataset/training_set/dogs/dog.1.jpg"

orb = cv2.ORB_create()
im = cv2.imread(image_path)
plt.imshow(im)


# Function for plotting keypoints
def draw_keypoints(vis, keypoints, color=(0, 255, 255)):
    for kp in keypoints:
        x, y = kp.pt
        plt.imshow(cv2.circle(vis, (int(x), int(y)), 2, color))


# Plotting the keypoints
kp = orb.detect(im, None)
kp, des = orb.compute(im, kp)
img = draw_keypoints(im, kp)
```
